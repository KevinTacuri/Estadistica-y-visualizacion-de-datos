##### 1. Realice un gráfico de línea de la función trigonométrica seno y coseno en dos gráficos distintos, cuyo dominio va de 0 a 2π, utilizando el método ‘subplot’.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0, 2 * np.pi, 0.1)
y1 = np.sin(x)
y2 = np.cos(x)

plt.subplot(2, 2, 1)
plt.title("Seno")
plt.plot(x, y1)
plt.grid()

plt.subplot(2, 2, 2)
plt.title("Coseno")
plt.plot(x, y2)
plt.grid()
```

##### 2. A partir del archivo ‘WEO_Data_Latinoamerica_2026.csv’, realice un gráfico de áreas del PBI del año 2024 al 2030 de los siguientes países: Brasil, México, Argentina, Colombia, Chile y Perú.

```python
import pandas as pd

FILE_NAME = "WEO_Data_Latinoamerica_2026.csv"
df = pd.read_csv(FILE_NAME, sep=",")
df.rename(columns={"COUNTRY": "País"}, inplace=True)
df.set_index("País", inplace=True)
df.head()
```

```python
años = list(map(str, range(2024, 2031)))
años
```

```python
paises = df.loc[["Brazil", "Mexico", "Argentina", "Colombia", "Chile", "Peru"], años]
paises = paises.T
paises.plot(kind="area", stacked=False)
plt.title("PBI comparado")
plt.xlabel("Años")
plt.ylabel("Mil Millones")
plt.show()
```

##### 3. A partir del archivo ‘WEO_Data_Latinoamerica_2026.csv’, grafique un histograma del número de países, cuyo PBI sea mayor a los 200 000 millones de dólares en el año 2021.

```python
pbi = df[df["2021"] > 200]["2021"]

pbi.plot.hist(bins=10)

plt.title("PBI de los países en 2021")
plt.xlabel("PBI en millones de dólares")
plt.ylabel("Número de países")
plt.show()
```

##### 4. A partir del archivo ‘WEO_Data_Latinoamerica_2026.csv’, realice un gráfico box plot del PBI del año 2020 al 2026 de los tres países con mayor PBI de Latinoamérica.

```python
import pandas as pd

FILE_NAME = "WEO_Data_Latinoamerica_2026.csv"
df = pd.read_csv(FILE_NAME, sep=",")
df.set_index("COUNTRY", inplace=True)
df.head()
```

```python
años = list(map(str, range(2020, 2027)))
años
```

```python
df["Promedio_2020_2026"] = df[años].mean(axis=1)
mayores_3 = df.nlargest(3, "Promedio_2020_2026")
```

```python
datos = mayores_3[años].T
datos.columns = mayores_3.index

datos.plot.box()

plt.title("PBI 2020-2026 de 3 países con mayor PBI")
plt.xlabel("País")
plt.ylabel("PBI en millones de dólares")
plt.show()
```
